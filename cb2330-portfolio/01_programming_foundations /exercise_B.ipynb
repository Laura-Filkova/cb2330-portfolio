Pick a formula from the table. Then in a new notebook:

| | Formula | Reads as |
|---|---|---|
| GC content | $\dfrac{1}{L}\sum_{i=1}^{L} \mathbb{1}[\,b_i \in \{G, C\}\,]$ | the fraction of a sequence that is G or C |
| Hamming distance | $\sum_{i=1}^{L} \mathbb{1}[\,a_i \neq b_i\,]$ | how many positions two equal-length sequences differ at |
| Running total | $S_k = \sum_{i=1}^{k} x_i$ for every $k$ | the total so far, one value per position |
| Dot product | $\sum_{i=1}^{N} a_i b_i$ | how much two vectors of numbers agree - and a **weighted** sum, which is a shape we need again later in the course |
| Geometric mean | $\left(\prod_{i=1}^{N} x_i\right)^{1/N}$ | an average for quantities that multiply, like fold changes |
| Harmonic mean | $N \Big/ \displaystyle\sum_{i=1}^{N} \frac{1}{x_i}$ | an average for rates |
| Binomial coefficient | $\dbinom{N}{k} = \dfrac{N!}{k!\,(N-k)!}$ | how many ways to choose $k$ things out of $N$ |
| Median | the middle value of the sorted data | the summary the mean disagreed with above |

Two of them have a $\prod$ rather than a $\sum$: same translation as section 3, with the
running total started at 1 and the `+` replaced by a `*`. The median has no summation sign
and has to be worked out from a sentence.




```python
import numpy as np
```

```python
#GC content without numpy
def count_gc(seq):
  total = 0
  for n in seq:
    if n == 'G' or n == 'C':
      total += 1
  return total/len(seq)

#GC content with numpy
def count_gc_numpy(seq):
  return (np.char.count(seq,'G') + np.char.count(seq,'C')) / len(seq)

```

```python
assert count_gc("ATGCTGACATG") == count_gc_numpy("ATGCTGACATG")
```

```python
#Hamming distance
def hamming_distance(seq1,seq2):
  total = 0
  for n1,n2 in zip(seq1,seq2):
    if n1 != n2:
      total += 1
  return total

#Hamming distance with numpy
def hamming_distance_numpy(seq1,seq2):
  return sum(np.array(list(seq1)) != np.array(list(seq2)))
```

```python
assert hamming_distance("ATGCTGACATG","AGGTTGACCTG") == hamming_distance_numpy("ATGCTGACATG","AGGTTGACCTG")
```

```python
#Running total
def running_total(values):
  total = 0
  for v in values:
    total += v
  return total

#Running total with numpy
def running_total_numpy(values):
  return sum(values)
```

```python
assert running_total([1,2,3,4,9,7,8]) == running_total_numpy([1,2,3,4,9,7,8])
```

```python
#dot product
def dot_product(a,b):
  total = 0
  for av,bv in zip(a,b):
    total += av*bv
  return total

#Running total with numpy
def dot_product_numpy(a,b):
  return sum(np.array(a)*np.array(b))
```

```python
assert dot_product([1,5,4,9,7,3],[6,4,8,7,2,3]) == dot_product_numpy([1,5,4,9,7,3],[6,4,8,7,2,3])
```

```python
#geometric mean
def geometric_mean(values):
  total = 1
  for v in values:
    total *= v
  return total**(1/len(values))

#geometric mean with numpy
def geometric_mean_numpy(values):
  return np.prod(values)**(1/len(values))
```

```python
assert geometric_mean([1,2,3,4,9,7,8]) == geometric_mean_numpy([1,2,3,4,9,7,8])
```

```python
#harmonic mean
def harmonic_mean(values):
  total = 0
  for v in values:
    total += 1/v
  return len(values)/total

#harmonic mean with numpy
def harmonic_mean_numpy(values):
  return len(values)/sum(1/np.array(values))
```

```python
assert harmonic_mean([1,2,3,4,9,7,8]) == harmonic_mean_numpy([1,2,3,4,9,7,8])
```

```python
#Binomial coeficient
def bin_coef(N,k):
  N_f = 1
  k_f = 1
  Nk_f = 1
  for i in range(1,N+1):
    N_f *= i
  for i in range(1,k+1):
    k_f *= i
  for i in range(1,N-k+1):
    Nk_f *= i
  return N_f/(k_f*Nk_f)

#Binomial coeficient with numpy
def bin_coef_numpy(N,k):
  return (np.prod(np.arange(1,N+1)))/(np.prod(np.arange(1,k+1))*np.prod(np.arange(1,N-k+1)))
```

```python
assert bin_coef(5,2) == bin_coef_numpy(5,2)
```

```python
#median
def median(values):
  values.sort()
  if len(values)%2 == 0:
    return (values[len(values)//2-1] + values[len(values)//2])/2
  else:
    return values[len(values)//2]
#median with numpy
def median_numpy(values):
  return np.median(values)
```

```python
assert median([1,2,3,4,9,7,8]) == median_numpy([1,2,3,4,9,7,8])
```
